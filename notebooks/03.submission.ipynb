```python
import pandas as pd
import joblib
train = pd.read_csv("../data/raw/train.csv").copy()
labels = train["Survived"]
train = train.drop("Survived", axis=1)
test = pd.read_csv("../data/raw/test.csv").copy()

model_randomforest = joblib.load("../data/processed/03/baseline_random_forest.pkl")
model_logistic_regression = joblib.load("../data/processed/03/baseline_logistic_regression.pkl")
```

```python
model_randomforest.fit(train, labels)

predict = model_randomforest.predict(test)

submission = pd.DataFrame({
    "PassengerId": test["PassengerId"].astype(int),
    "Survived": predict.astype(int)
})
submission = submission.sort_values("PassengerId")
submission.to_csv("../data/processed/04/submission.csv", index=False)

print(submission.shape)   # (418, 2) 여야 함
print(submission.head())
```
